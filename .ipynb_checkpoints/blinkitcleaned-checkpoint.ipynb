```python
import pandas as pd
import os

pd.set_option('display.max_columns', None)

RAW = "./"
OUT = "./clean/"
os.makedirs(OUT, exist_ok=True)

print("Ready. Files in this folder:")
print([f for f in os.listdir(RAW) if f.endswith('.csv')])
```

```python
files = {
    'customers': 'blinkit_customers.csv',
    'orders': 'blinkit_orders.csv',
    'order_items': 'blinkit_order_items.csv',
    'products': 'blinkit_products.csv',
    'delivery': 'blinkit_delivery_performance.csv',
    'feedback': 'blinkit_customer_feedback.csv',
    'marketing': 'blinkit_marketing_performance.csv',
    'inventory': 'blinkit_inventory.csv',
}

for name, f in files.items():
    df = pd.read_csv(RAW + f)
    print(f'--- {name} ({df.shape[0]} rows, {df.shape[1]} cols) ---')
    nulls = df.isnull().sum()
    nulls = nulls[nulls > 0]
    if len(nulls):
        print('nulls:', dict(nulls))
    dups = df.duplicated().sum()
    if dups:
        print('duplicate rows:', dups)
    print()
```

```python
customers = pd.read_csv(RAW + "blinkit_customers.csv")
customers["registration_date"] = pd.to_datetime(customers["registration_date"])
customers.to_csv(OUT + "customers.csv", index=False)
customers.head()
```

```python
products = pd.read_csv(RAW + "blinkit_products.csv")
products.to_csv(OUT + "products.csv", index=False)
products.head()
```

```python
orders = pd.read_csv(RAW + "blinkit_orders.csv")
for col in ["order_date", "promised_delivery_time", "actual_delivery_time"]:
    orders[col] = pd.to_datetime(orders[col])
orders.to_csv(OUT + "orders.csv", index=False)
orders.head()
```

```python
order_items = pd.read_csv(RAW + "blinkit_order_items.csv")
order_items["line_total"] = order_items["quantity"] * order_items["unit_price"]
order_items.to_csv(OUT + "order_items.csv", index=False)
order_items.head()
```

```python
delivery = pd.read_csv(RAW + "blinkit_delivery_performance.csv")
for col in ["promised_time", "actual_time"]:
    delivery[col] = pd.to_datetime(delivery[col])
delivery["reasons_if_delayed"] = delivery["reasons_if_delayed"].fillna("Not Delayed")
delivery.to_csv(OUT + "delivery_performance.csv", index=False)
delivery.head()
```

```python
feedback = pd.read_csv(RAW + "blinkit_customer_feedback.csv")
feedback["feedback_date"] = pd.to_datetime(feedback["feedback_date"])
feedback.to_csv(OUT + "customer_feedback.csv", index=False)
feedback.head()
```

```python
marketing = pd.read_csv(RAW + "blinkit_marketing_performance.csv")
marketing["date"] = pd.to_datetime(marketing["date"])
marketing.to_csv(OUT + "marketing_performance.csv", index=False)
marketing.head()
```

```python
inventory = pd.read_csv(RAW + "blinkit_inventory.csv")
inventory["date"] = pd.to_datetime(inventory["date"], dayfirst=True)
inventory.to_csv(OUT + "inventory.csv", index=False)
inventory.head()
```

```python
cleaned = ["customers", "products", "orders", "order_items",
           "delivery_performance", "customer_feedback",
           "marketing_performance", "inventory"]

for f in cleaned:
    df = pd.read_csv(OUT + f + ".csv")
    print(f"{f}: {df.shape[0]} rows, {df.shape[1]} cols")
```
